````python
# BMI calculator - a small clinical example of if/elif/else
# Interactive version:
#   weight = float(input("Enter your weight in kg: "))
#   height = float(input("Enter your height in meters: "))
weight = 68.0       # kg
height = 1.72       # meters

# Calculate BMI
bmi = weight / (height ** 2)

print(f"BMI: {bmi:.2f}")

# Classify BMI (WHO cut-offs)
if bmi < 18.5:
    print("Underweight")
elif bmi < 24.9:
    print("Normal weight")
elif bmi < 29.9:
    print("Overweight")
else:
    print("Obese")
````

## 1.30 Putting Part 1 Together: A DNA Toolkit

Nothing new is introduced here. This uses only strings, dictionaries, loops, conditionals,
comprehensions and functions from the sections above - and it is very close to what a real
first bioinformatics script looks like: count bases, compute GC content, transcribe to RNA,
take the reverse complement, and read codons.

````python
# A short stretch of DNA, stored as a string
dna = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"
````

````python
# 1. Length of the sequence -> len() works on any string
print("Length          :", len(dna), "bases")
````

````python
# 2. Count each base -> a dictionary built inside a for loop
counts = {}
for base in dna:
    if base in counts:
        counts[base] = counts[base] + 1
    else:
        counts[base] = 1
print("Base counts     :", counts)
````

````python
# 3. GC content -> arithmetic wrapped in a reusable function
def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    gc = sequence.count("G") + sequence.count("C")
    return 100 * gc / len(sequence)

print("GC content      :", f"{gc_content(dna):.1f}%")
````

````python
# 4. Transcription DNA -> RNA: every T becomes U (a string method)
rna = dna.replace("T", "U")
print("RNA             :", rna)
````

````python
# 5. Reverse complement -> dictionary lookup over a reversed string
complement = {"A": "T", "T": "A", "G": "C", "C": "G"}
rev_comp = "".join(complement[base] for base in reversed(dna))
print("Rev. complement :", rev_comp)
````

````python
# 6. Split into codons (triplets) -> slicing with a step
codons = [dna[i:i + 3] for i in range(0, len(dna) - 2, 3)]
print("codons  :", codons)
````

````python
# 7. A tiny genetic code -> dictionary plus a conditional
genetic_code = {"ATG": "Met", "GCC": "Ala", "ATT": "Ile", "TGA": "STOP"}
for codon in codons:
    if codon in genetic_code:
        print(f"  {codon} -> {genetic_code[codon]}")
    else:
        print(f"  {codon} -> not in our mini table")
````

# Part 2 - Scientific Computing with NumPy

Part 1 gave you Python lists. Lists are flexible but slow, and they cannot do arithmetic:
`[1, 2, 3] * 2` repeats the list instead of doubling the numbers. Biology produces numbers in
bulk - 20,000 gene expression values per sample, a 512x512 microscopy image, a distance matrix
between 300 sequences - and you need arithmetic on all of them at once.

That is what NumPy provides. Its `ndarray` stores one data type in one contiguous block of
memory, so an operation on a million values happens in compiled C, not in a Python loop:

```python
expression = np.array([12.4, 3.1, 88.0, 0.5])
log_expr = np.log2(expression + 1)     # one line, whole array, no loop
```

This is called **vectorisation**, and it is the single most important habit in scientific
Python. Every library ahead of you - Pandas, scikit-learn, PyTorch - is built on arrays.

**Sections:** what an array is, creating arrays, arithmetic, indexing, aggregation, matrices,
the dot product, matrix indexing and aggregation, transposing, reshaping, N-D arrays.

## 3.9 Missing Values

Real datasets have holes: a failed assay, an unrecorded measurement, a column that only applies
to some samples. Pandas marks these as `NaN` (Not a Number) and every summary function skips
them by default. You must decide *deliberately* what to do with them, because most
machine-learning models refuse to accept `NaN` at all.

Three standard moves:

- **Report** them: `df.isna().sum()`
- **Drop** them: `df.dropna()` - safe when few rows are affected, wasteful otherwise
- **Fill** them (impute): `df.fillna(value)` - commonly the column mean or median

`ecoli.csv` happens to be complete, so we punch a couple of holes in a copy to practise on.

````python
# Work on a copy so the original stays clean
df_missing = df.copy()

# Introduce missing values on purpose, in rows 0-2 of one column
df_missing.loc[0:2, "mcg"] = np.nan

# 1. Report
print("Missing per column:")
print(df_missing.isna().sum().head())

# 2. Drop the affected rows
print("\nShape before dropna:", df_missing.shape)
print("Shape after  dropna:", df_missing.dropna().shape)

# 3. Fill with the column mean (a simple imputation)
filled = df_missing.fillna({"mcg": df_missing["mcg"].mean()})
print("\nMissing after fillna:", filled["mcg"].isna().sum())
filled.head(3)
````

## 4.5 Choose Your Plot
Matplotlib offers several kind of plots ([see Gallery](https://matplotlib.org/stable/gallery/index.html))


### Scatter Plot
A scatterplot displays the relationship between 2 `numeric variables`. For each data point, the value of its first variable is represented on the X axis, the second on the Y axis.

- `X = np.random.uniform(0, 1, 100):` Generates an array of 100 random numbers between 0 and 1 for the x-axis.
- `Y = np.random.uniform(0, 1, 100):` Generates an array of 100 random numbers between 0 and 1 for the y-axis.
- `plt.scatter(X, Y):` Creates a scatter plot using the generated random data. Each point in the plot corresponds to a pair of values from the X and Y arrays.
- `plt.show():` Displays the scatter plot.

````python
X = np.random.uniform(0, 1, 100)
Y = np.random.uniform(0, 1, 100)
plt.scatter(X, Y)

plt.show()
````

### Bar Plot
A barplot shows the relationship `between a numeric` and a categoric variable. Each entity of the categoric variable is represented as a bar. The size of the bar represents its numeric value.

- `X = np.arange(10):` Creates an array with values from 0 to 9. This will be used as the x-axis values for the bar plot.
- `Y = np.random.uniform(1, 10, 10):` Generates an array of 10 random numbers between 1 and 10. Each y-value corresponds to a bar in the plot.
- `plt.bar(X, Y):` Creates a bar plot using the generated x and y values. The x-values are taken from the array X, and the y-values are taken from the array Y.
- `plt.show():` Displays the bar plot.


````python
X = np.arange(10)
Y = np.random.uniform(1, 10, 10)
plt.bar(X, Y)
plt.show()
````

An image plot is a type of visualization that represents a 2D array of data as a graphical image. In the context of Matplotlib, the imshow function is specifically designed for creating image plots.

- `Z = np.random.uniform(0, 1, (8, 8)):` Creates an 8x8 array of random numbers between 0 and 1. This 2D array serves as the data for the image plot.
- `plt.imshow(Z):` Creates an image plot using the 2D array Z. Each element in the array corresponds to a pixel in the image.
- `plt.show(): `Displays the image plot.

## 5.1 The Seaborn Mental Model

Two things to understand before the plots.

**1. Tidy data.** Seaborn expects a *long* / tidy table: one row per observation, one column per
variable. You then map columns onto visual roles:

| Argument | Role |
|:--|:--|
| `data` | The DataFrame |
| `x`, `y` | Position |
| `hue` | Colour - split by a category |
| `size`, `style` | Marker size and shape |
| `col`, `row` | Split into separate panels |

**2. Figure-level vs axes-level functions.** This distinction confuses everyone once, so learn
it now:

| Kind | Examples | Returns | Use when |
|:--|:--|:--|:--|
| **Axes-level** | `scatterplot`, `boxplot`, `histplot`, `heatmap` | a Matplotlib `Axes` | you are placing a plot inside your own `fig, ax` layout |
| **Figure-level** | `relplot`, `catplot`, `displot`, `lmplot`, `pairplot` | a Seaborn `FacetGrid` | you want automatic multi-panel splits with `col=` / `row=` |

Rule of thumb: **axes-level by default; figure-level when you need facets.** A figure-level
function creates its own figure, so `ax=` does not work with it - that is the error people hit.

````python
# Seaborn is already imported and themed in the Setup cell, but here it is again explicitly
import seaborn as sns

print("Seaborn version:", sns.__version__)

# The two datasets for this part.
# 1. E. coli protein localisation (7 numeric features + a categorical target)
ecoli = pd.read_csv(os.path.join(DATA_DIR, "ecoli.csv"))

# 2. Iris flowers. The file has no header row, so we supply the column names ourselves.
iris = pd.read_csv(
    os.path.join(DATA_DIR, "iris.csv"),
    header=None,
    names=["sepal_length", "sepal_width", "petal_length", "petal_width", "species"],
)
# Tidy up the labels: "Iris-setosa" -> "setosa"
iris["species"] = iris["species"].str.replace("Iris-", "", regex=False)

print("\necoli:", ecoli.shape, "| iris:", iris.shape)
iris.head()
````

## 5.2 Distributions - What Does One Variable Look Like?

Always the first question about a numeric column. `histplot` counts observations in bins;
`kdeplot` draws a smoothed density; `ecdfplot` shows the cumulative proportion and needs no bin
choice at all. Add `hue` and Seaborn overlays one distribution per group - the fastest way to
see whether a feature separates your classes.

````python
# Axes-level: three views of the same variable, side by side in our own figure
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(data=iris, x="petal_length", bins=20, kde=True, ax=axes[0])
axes[0].set_title("histplot: counts per bin (+ KDE)")

sns.kdeplot(data=iris, x="petal_length", hue="species", fill=True, alpha=0.4, ax=axes[1])
axes[1].set_title("kdeplot with hue: one density per species")

sns.ecdfplot(data=iris, x="petal_length", hue="species", ax=axes[2])
axes[2].set_title("ecdfplot: cumulative proportion")

fig.tight_layout()
plt.show()
````

The middle panel makes the biology obvious: *setosa* petals are completely separated
from the other two species, while *versicolor* and *virginica* overlap. A classifier will find
setosa trivial and the other two harder - and you learned that from one line of plotting code,
before training anything.

The same check on our protein data explains the bimodal histogram you drew in Part 4:

````python
# Figure-level: displot splits into its own panels with col=
g = sns.displot(
    data=ecoli,
    x="alm1",
    hue="site",
    col="site",
    col_wrap=4,
    height=2.6,
    bins=15,
    facet_kws={"sharey": False},
    legend=False,
)
g.set_titles("site = {col_name}")
g.set_axis_labels("ALOM membrane score (alm1)", "Count")
g.figure.suptitle("The bimodality is the inner membrane (im) class, not noise", y=1.04)
plt.show()
````

## 5.3 Categories - How Does a Number Differ Between Groups?

The everyday question in biology: does this measurement differ between compartments, treatments,
genotypes, or patient groups? Seaborn's categorical functions all take `x` = category,
`y` = number.

| Function | Shows | Best for |
|:--|:--|:--|
| `countplot` | How many rows per category | Class balance |
| `boxplot` | Median, quartiles, outliers | Compact comparison across many groups |
| `violinplot` | Full density, mirrored | Revealing bimodality a box hides |
| `stripplot` / `swarmplot` | Every individual point | Small n, where showing raw data is honest |
| `barplot` | Mean plus a 95% bootstrap confidence interval | Summary with uncertainty |
| `pointplot` | Means joined by lines | Trends across an ordered category |

**A box plot can lie.** Two very different distributions can share the same five-number summary,
so with fewer than ~50 points per group, overlay the raw data.

````python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# 1. How many proteins per compartment?
sns.countplot(data=ecoli, x="site", ax=axes[0, 0], hue="site", legend=False)
axes[0, 0].set_title("countplot: class balance (cp and im dominate)")

# 2. Median and spread of one feature across compartments
sns.boxplot(data=ecoli, x="site", y="mcg", ax=axes[0, 1], hue="site", legend=False)
axes[0, 1].set_title("boxplot: mcg is highest in pp and om")

# 3. Full densities - note the shapes a box plot would hide
sns.violinplot(data=ecoli, x="site", y="alm1", ax=axes[1, 0], hue="site", legend=False)
axes[1, 0].set_title("violinplot: alm1 is bimodal within im")

# 4. Box plus every raw point: the honest default for small groups
sns.boxplot(data=ecoli, x="site", y="aac", ax=axes[1, 1], color="lightgray",
            showfliers=False)
sns.stripplot(data=ecoli, x="site", y="aac", ax=axes[1, 1], size=3,
              color="black", alpha=0.5, jitter=0.25)
axes[1, 1].set_title("boxplot + stripplot: summary AND raw data")

fig.tight_layout()
plt.show()
````

````python
# barplot draws the MEAN with a 95% bootstrap confidence interval - the error bar is
# the uncertainty in the mean, not the spread of the data. Say which one you mean in captions.
fig, ax = plt.subplots(figsize=(7, 4))

sns.barplot(data=ecoli, x="site", y="gvh", errorbar=("ci", 95), ax=ax,
            hue="site", legend=False)
ax.set_xlabel("Localisation site")
ax.set_ylabel("Mean von Heijne score (gvh)")
ax.set_title("Mean gvh per compartment, with 95% CI")
plt.show()
````

````python
# lineplot: ordered x, repeated measurements. Here is a simulated growth curve -
# 3 strains x 3 replicates, the shape of a real plate-reader experiment.
hours = np.arange(0, 13)
records = []
for strain, rate in [("wild-type", 0.45), ("mutant A", 0.30), ("mutant B", 0.18)]:
    for replicate in range(3):
        od = 0.05 * np.exp(rate * hours) + rng.normal(0, 0.05, size=hours.size)
        records.append(pd.DataFrame({"hour": hours, "OD600": od,
                                     "strain": strain, "replicate": replicate}))
growth = pd.concat(records, ignore_index=True)

fig, ax = plt.subplots(figsize=(7, 4.5))
sns.lineplot(data=growth, x="hour", y="OD600", hue="strain", marker="o", ax=ax)
ax.set_xlabel("Time (hours)")
ax.set_ylabel("Optical density (OD600)")
ax.set_title("lineplot averages the replicates and shades the 95% CI automatically")
plt.show()

growth.head(3)
````

## 5.6 Matrices - Correlation and Clustering

`heatmap` draws a 2-D table as colour. Its most common use is a correlation matrix: which
features carry the same information? Strongly correlated features are redundant, and dropping
one often makes a model simpler with no loss of accuracy.

`clustermap` goes further, reordering rows and columns by hierarchical clustering - the standard
figure for gene expression data, where the clustering itself is the result.

````python
features = ["mcg", "gvh", "lip", "chg", "aac", "alm1", "alm2"]
corr = ecoli[features].corr()

fig, ax = plt.subplots(figsize=(6.5, 5.5))
sns.heatmap(
    corr,
    annot=True,          # print the number in each cell
    fmt=".2f",
    cmap="vlag",         # diverging: red / blue around zero - correct for correlations
    center=0,            # anchor the colour scale at 0
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5,
    cbar_kws={"label": "Pearson r"},
    ax=ax,
)
ax.set_title("Feature correlations: alm1 and alm2 are near-duplicates")
plt.show()
````

````python
# clustermap: z-score each feature, then cluster proteins and features together.
# Row colours show the true localisation site, so you can see whether the
# clustering recovers the biology WITHOUT being told the labels.
site_palette = dict(zip(sorted(ecoli["site"].unique()),
                        sns.color_palette("tab10", ecoli["site"].nunique())))
row_colors = ecoli["site"].map(site_palette)

g = sns.clustermap(
    ecoli[features],
    z_score=1,            # standardise each feature (column)
    cmap="vlag",
    center=0,
    row_colors=row_colors,
    figsize=(7, 7),
    yticklabels=False,
    cbar_kws={"label": "z-score"},
)
g.figure.suptitle("clustermap: unsupervised structure vs. the true labels (row colours)", y=1.01)
plt.show()
````

## 5.7 Themes, Palettes and Context

Seaborn's second job is making plots look right with no fiddling.

- `sns.set_theme(style=...)` - `"whitegrid"`, `"darkgrid"`, `"white"`, `"ticks"`, `"dark"`
- `sns.set_context(...)` - `"paper"`, `"notebook"`, `"talk"`, `"poster"`; scales fonts and lines
  so one figure works both in a thesis and on a projector
- `sns.despine()` - remove the top and right frame lines (a small, standard improvement)

**Choosing a palette is a correctness issue, not decoration:**

| Data | Palette type | Examples |
|:--|:--|:--|
| Unordered categories | Qualitative | `deep`, `colorblind`, `tab10`, `Set2` |
| Low to high | Sequential | `viridis`, `mako`, `rocket`, `Blues` |
| Diverging around a midpoint (correlation, log fold change) | Diverging | `vlag`, `coolwarm`, `RdBu_r` |

Prefer `colorblind` or `viridis`: roughly 1 in 12 men has a colour-vision deficiency, and the
old `jet` rainbow invents boundaries that are not in the data. Never encode a *number* with a
qualitative palette, or a *category* with a sequential one.

````python
# The same plot under four styles
styles = ["whitegrid", "darkgrid", "white", "ticks"]
fig, axes = plt.subplots(1, 4, figsize=(16, 3.4))

for style, ax in zip(styles, axes):
    with sns.axes_style(style):                     # temporary style, scoped to this block
        sns.scatterplot(data=iris, x="sepal_length", y="sepal_width",
                        hue="species", legend=False, ax=ax)
        ax.set_title(f'style="{style}"')

fig.tight_layout()
plt.show()
````

````python
# Palette types, and what goes wrong if you mismatch them
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Qualitative for unordered categories - correct
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species",
                palette="colorblind", ax=axes[0])
axes[0].set_title("Qualitative palette for categories (correct)")

# Sequential for a number - correct
sns.scatterplot(data=ecoli, x="mcg", y="gvh", hue="alm1",
                palette="viridis", ax=axes[1])
axes[1].set_title("Sequential palette for a number (correct)")

# Qualitative for a number - WRONG: the colours imply groups that do not exist
sns.scatterplot(data=ecoli, x="mcg", y="gvh", hue="alm1",
                palette="Set1", legend=False, ax=axes[2])
axes[2].set_title("Qualitative palette for a number (misleading)")

fig.tight_layout()
plt.show()
````

## 5.8 Combining Seaborn with Matplotlib

The pay-off of learning Part 4 first: a Seaborn axes-level call returns a plain Matplotlib
`Axes`, so you can keep going with every method you already know - `set_xlabel`, `axhline`,
`annotate`, `set_ylim`, `savefig`. Seaborn does the statistics; Matplotlib does the polish.

This is the pattern for a figure you would actually submit.

````python
sns.set_theme(style="ticks", context="talk")

fig, ax = plt.subplots(figsize=(9, 5.5))

# Seaborn: the statistical layer
sns.violinplot(data=ecoli, x="site", y="mcg", ax=ax,
               hue="site", palette="colorblind", legend=False, inner=None, alpha=0.55)
sns.stripplot(data=ecoli, x="site", y="mcg", ax=ax,
              color="black", size=2.5, alpha=0.5, jitter=0.22)

# Matplotlib: the polish
overall_mean = ecoli["mcg"].mean()
ax.axhline(overall_mean, color="crimson", linestyle="--", linewidth=1.5, zorder=0)
ax.annotate(
    f"overall mean = {overall_mean:.2f}",
    xy=(0.99, overall_mean),
    xycoords=("axes fraction", "data"),
    xytext=(0, 8),
    textcoords="offset points",
    ha="right",
    color="crimson",
    fontsize=11,
)
ax.set_xlabel("Subcellular localisation site")
ax.set_ylabel("McGeoch signal score (mcg)")
ax.set_title("Signal-peptide score by compartment", pad=12)
ax.set_ylim(0, 1.05)
sns.despine(trim=True)
fig.tight_layout()

fig.savefig("ecoli_mcg_by_site.png", dpi=300, bbox_inches="tight")
plt.show()

# Reset the theme for the rest of the notebook
sns.set_theme(style="whitegrid", context="notebook")
````

## The Habits That Matter More Than the Syntax

1. **Look at the data before modelling.** `head`, `info`, `describe`, `value_counts`, then a
   `pairplot`. Most catastrophic modelling mistakes are visible in those five commands.
2. **Vectorise.** If you are writing a `for` loop over rows of a DataFrame, there is almost
   certainly a one-line array or `groupby` version that is both faster and clearer.
3. **Check class balance.** The *E. coli* dataset has 143 `cp` proteins and 2 `imL`. A model that
   always predicts `cp` scores 43% accuracy while learning nothing.
4. **Label every axis.** A figure a reader cannot interpret unaided is not a result.
5. **Choose palettes by data type.** Qualitative for categories, sequential for magnitude,
   diverging around a meaningful midpoint. Prefer colourblind-safe options.
6. **Make it reproducible.** Fix your random seed, keep raw data read-only, write cleaned data to
   a new file, and run your notebook top-to-bottom before you trust it.
